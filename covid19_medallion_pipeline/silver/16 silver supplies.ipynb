```python
# ============================================================
# Pipeline : covid19_medallion_pipeline
# Notebook : 04_silver/16_silver_supplies
# ============================================================

import dlt
from pyspark.sql import functions as F


@dlt.table(name="covid19_lakehouse.silver.supplies", comment="Supplies with correct types")
@dlt.expect_all_or_drop({
    "valid_patient_id": "patient_id IS NOT NULL",
    "valid_encounter_id": "encounter_id IS NOT NULL",
})
def silver_supplies():
    df = dlt.read_stream("covid19_lakehouse.bronze.supplies")
    return (
        df.select(
            F.col("PATIENT").alias("patient_id"),
            F.col("ENCOUNTER").alias("encounter_id"),
            F.to_date("DATE").alias("supply_date"),
            F.col("CODE").alias("supply_code"),
            F.col("DESCRIPTION").alias("supply_description"),
            F.col("QUANTITY").cast("int").alias("quantity"),
        )
    )
```
