```python
# ============================================================
# Pipeline : covid19_medallion_pipeline
# Notebook : 04_silver/15_silver_immunizations
# ============================================================

import dlt
from pyspark.sql import functions as F


@dlt.table(name="covid19_lakehouse.silver.immunizations", comment="Immunizations with correct types")
@dlt.expect_all_or_drop({
    "valid_patient_id": "patient_id IS NOT NULL",
    "valid_encounter_id": "encounter_id IS NOT NULL",
})
def silver_immunizations():
    df = dlt.read_stream("covid19_lakehouse.bronze.immunizations")
    return (
        df.select(
            F.col("PATIENT").alias("patient_id"),
            F.col("ENCOUNTER").alias("encounter_id"),
            F.to_date("DATE").alias("immunization_date"),
            F.col("CODE").alias("immunization_code"),
            F.col("DESCRIPTION").alias("immunization_description"),
            F.col("BASE_COST").cast("decimal(10,2)").alias("base_cost"),
        )
    )
```
