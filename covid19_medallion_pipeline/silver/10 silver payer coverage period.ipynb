```python
# ============================================================
# Pipeline : covid19_medallion_pipeline
# Notebook : 04_silver/10_silver_payer_coverage_period
# ============================================================

import dlt
from pyspark.sql import functions as F


@dlt.table(name="covid19_lakehouse.silver.payer_coverage_period", comment="Patient-payer coverage periods")
@dlt.expect_all_or_drop({
    "valid_patient_id": "patient_id IS NOT NULL",
    "valid_payer_id": "payer_id IS NOT NULL",
})
def silver_payer_coverage_period():
    df = dlt.read_stream("covid19_lakehouse.bronze.payer_transitions")
    return (
        df.select(
            F.col("PATIENT").alias("patient_id"),
            F.col("PAYER").alias("payer_id"),
            F.col("START_YEAR").cast("int").alias("start_year"),
            F.col("END_YEAR").cast("int").alias("end_year"),
            F.col("OWNERSHIP").alias("ownership"),
            (F.col("END_YEAR").cast("int") - F.col("START_YEAR").cast("int")).alias("duration_years"),
        )
    )
```
