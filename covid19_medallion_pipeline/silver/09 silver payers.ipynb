```python
# ============================================================
# Pipeline : covid19_medallion_pipeline
# Notebook : 04_silver/09_silver_payers
# ============================================================

import dlt
from pyspark.sql import functions as F


@dlt.table(name="covid19_lakehouse.silver.payers", comment="Payers with correct types")
def silver_payers():
    df = dlt.read_stream("covid19_lakehouse.bronze.payers")
    return (
        df.select(
            F.col("Id").alias("payer_id"),
            F.col("NAME").alias("payer_name"),
            F.col("ADDRESS").alias("address"),
            F.col("CITY").alias("city"),
            F.col("STATE_HEADQUARTERED").alias("state_headquartered"),
            F.col("ZIP").alias("zip"),
            F.col("PHONE").alias("phone"),
            F.col("_ingested_at"),
        )
    )
```
