```python
# ============================================================
# Pipeline : covid19_medallion_pipeline
# Notebook : 04_silver/06_silver_procedures
# ============================================================

import dlt
from pyspark.sql import functions as F


@dlt.table(name="covid19_lakehouse.silver.procedures", comment="Procedures with correct types")
@dlt.expect_all_or_drop({
    "valid_patient_id": "patient_id IS NOT NULL",
    "valid_encounter_id": "encounter_id IS NOT NULL",
})
def silver_procedures():
    df = dlt.read_stream("covid19_lakehouse.bronze.procedures")
    return (
        df.select(
            F.col("PATIENT").alias("patient_id"),
            F.col("ENCOUNTER").alias("encounter_id"),
            F.to_date("DATE").alias("procedure_date"),
            F.col("CODE").alias("procedure_code"),
            F.col("DESCRIPTION").alias("procedure_description"),
            F.col("BASE_COST").cast("decimal(10,2)").alias("base_cost"),
            F.col("REASONCODE").alias("reason_code"),
            F.col("REASONDESCRIPTION").alias("reason_description"),
            F.col("_ingested_at"),
        )
    )
```
