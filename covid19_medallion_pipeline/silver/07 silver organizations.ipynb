```python
# ============================================================
# Pipeline : covid19_medallion_pipeline
# Notebook : 04_silver/07_silver_organizations
# ============================================================

import dlt
from pyspark.sql import functions as F


@dlt.table(name="covid19_lakehouse.silver.organizations", comment="Organizations with correct types")
def silver_organizations():
    df = dlt.read_stream("covid19_lakehouse.bronze.organizations")
    return (
        df.select(
            F.col("Id").alias("organization_id"),
            F.col("NAME").alias("organization_name"),
            F.col("ADDRESS").alias("address"),
            F.col("CITY").alias("city"),
            F.col("STATE").alias("state"),
            F.col("ZIP").alias("zip"),
            F.col("LAT").cast("double").alias("lat"),
            F.col("LON").cast("double").alias("lon"),
            F.col("PHONE").alias("phone"),
            F.col("REVENUE").cast("decimal(15,2)").alias("revenue"),
            F.col("UTILIZATION").cast("bigint").alias("utilization"),
            F.col("_ingested_at"),
        )
    )
```
