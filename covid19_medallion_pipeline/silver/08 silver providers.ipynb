```python
# ============================================================
# Pipeline : covid19_medallion_pipeline
# Notebook : 04_silver/08_silver_providers
# ============================================================

import dlt
from pyspark.sql import functions as F


@dlt.table(name="covid19_lakehouse.silver.providers", comment="Providers with correct types")
def silver_providers():
    df = dlt.read_stream("covid19_lakehouse.bronze.providers")
    return (
        df.select(
            F.col("Id").alias("provider_id"),
            F.col("ORGANIZATION").alias("organization_id"),
            F.col("NAME").alias("provider_name"),
            F.col("GENDER").alias("gender"),
            F.col("SPECIALITY").alias("speciality"),
            F.col("ADDRESS").alias("address"),
            F.col("CITY").alias("city"),
            F.col("STATE").alias("state"),
            F.col("ZIP").alias("zip"),
            F.col("LAT").cast("double").alias("lat"),
            F.col("LON").cast("double").alias("lon"),
            F.col("UTILIZATION").cast("bigint").alias("utilization"),
            F.col("_ingested_at"),  # Required for apply_changes(sequence_by=...) in Gold
        )
    )
```
