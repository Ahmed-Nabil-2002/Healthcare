```python
# ============================================================
# Pipeline : covid19_medallion_pipeline
# Notebook : 04_silver/14_silver_imaging_studies
# ============================================================

import dlt
from pyspark.sql import functions as F


@dlt.table(name="covid19_lakehouse.silver.imaging_studies", comment="Imaging studies with correct types")
@dlt.expect_all_or_drop({
    "valid_patient_id": "patient_id IS NOT NULL",
    "valid_encounter_id": "encounter_id IS NOT NULL",
})
def silver_imaging_studies():
    df = dlt.read_stream("covid19_lakehouse.bronze.imaging_studies")
    return (
        df.select(
            F.col("Id").alias("imaging_study_id"),
            F.col("PATIENT").alias("patient_id"),
            F.col("ENCOUNTER").alias("encounter_id"),
            F.to_date("DATE").alias("study_date"),
            F.col("BODYSITE_CODE").alias("bodysite_code"),
            F.col("BODYSITE_DESCRIPTION").alias("bodysite_description"),
            F.col("MODALITY_CODE").alias("modality_code"),
            F.col("MODALITY_DESCRIPTION").alias("modality_description"),
            F.col("SOP_CODE").alias("sop_code"),
            F.col("SOP_DESCRIPTION").alias("sop_description"),
        )
    )
```
