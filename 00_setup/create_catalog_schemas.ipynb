```sql
-- ============================================================
-- Notebook: 00_setup/01_create_catalog_schemas
-- Project : COVID-19 Clinical Data Lakehouse
-- Purpose : Create the complete Unity Catalog structure (Catalog + Schemas + Volume)
--           Idempotently - Safe to rerun at any time without raising errors
-- ============================================================

-- 1) Main Project Catalog
CREATE CATALOG IF NOT EXISTS covid19_lakehouse
COMMENT 'COVID-19 Synthetic Clinical Data Lakehouse - Synthea 100k patients dataset';

-- 2) The Four Schemas (Medallion Architecture + Quarantine)
CREATE SCHEMA IF NOT EXISTS covid19_lakehouse.bronze
COMMENT 'Raw ingested data - no transformation applied';

CREATE SCHEMA IF NOT EXISTS covid19_lakehouse.silver
COMMENT 'Cleaned data with Data Quality rules applied';

CREATE SCHEMA IF NOT EXISTS covid19_lakehouse.gold
COMMENT 'Star Schema - Fact and Dimension tables for analytics';

CREATE SCHEMA IF NOT EXISTS covid19_lakehouse.quarantine
COMMENT 'Records that failed Silver layer DQ validation, kept for audit and DQ trend analysis';

-- 3) Volume for uploading Raw CSV files (Landing Zone prior to Auto Loader)
CREATE VOLUME IF NOT EXISTS covid19_lakehouse.bronze.raw_files
COMMENT 'Landing zone for raw Synthea CSV files, organized by source table subfolder';

-- 4) Verify that all catalog objects were created successfully
SHOW SCHEMAS IN covid19_lakehouse;
```
