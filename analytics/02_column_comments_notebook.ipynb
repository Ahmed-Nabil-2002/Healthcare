```sql

-- ============================================================
-- Notebook : 07_analytics/02_column_comments
-- Project  : COVID-19 Clinical Data Lakehouse
-- Purpose  : Adds table-level and column-level COMMENTs for every Gold
--            table (18 tables / 135 columns total) in ONE consolidated
--            notebook. This is the semantic layer that lets Genie AI
--           understand column meaning and table relationships (Foreign
--            Keys) before answering natural-language questions.
--
-- Structure: One notebook cell per table (COMMAND separators below),
--            so each table's comments can be reviewed/re-run
--            independently while staying in a single notebook.
--
-- Note     : All statements are idempotent - safe to re-run.
-- ============================================================
```
